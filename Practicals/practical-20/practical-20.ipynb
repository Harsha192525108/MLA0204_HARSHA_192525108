```python

# Future Sales Prediction using Linear Regression

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Create historical sales dataset
data = {
    'Month': np.arange(1, 13),

    'Sales': [100, 120, 130, 150, 160, 180,
              200, 210, 230, 250, 270, 300]
}

df = pd.DataFrame(data)

# Separate input and output
X = df[['Month']]
y = df['Sales']

# Use first 9 months for training
X_train = X.iloc[:9]
y_train = y.iloc[:9]

# Use last 3 months for testing
X_test = X.iloc[9:]
y_test = y.iloc[9:]

# Create Linear Regression model
model = LinearRegression()

# Train model
model.fit(X_train, y_train)

# Predict test sales
y_pred = model.predict(X_test)

# Evaluate model
print("FUTURE SALES PREDICTION")
print("-----------------------")

print("Actual Sales:", list(y_test))
print("Predicted Sales:", y_pred.round(2))

print("\nMean Absolute Error:",
      mean_absolute_error(y_test, y_pred))

print("Mean Squared Error:",
      mean_squared_error(y_test, y_pred))

print("R2 Score:", r2_score(y_test, y_pred))

# Predict future sales for next 3 months
future_months = pd.DataFrame({
    'Month': [13, 14, 15]
})

future_sales = model.predict(future_months)

print("\nFUTURE SALES FORECAST")

for month, sales in zip(
    future_months['Month'], future_sales
):
    print("Month", month, "Predicted Sales:",
          round(sales, 2))

# Plot historical and predicted sales
plt.figure(figsize=(8, 5))

plt.scatter(X, y, color='blue',
            label='Historical Sales')

plt.plot(X, model.predict(X), color='red',
         label='Regression Line')

plt.scatter(future_months['Month'], future_sales,
            color='green', label='Future Predictions')

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Future Sales Prediction using Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```
